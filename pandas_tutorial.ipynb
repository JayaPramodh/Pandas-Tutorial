## Accessing data with Pandas

```python
# read data from coffee.csv
coffee = pd.read_csv('coffee.csv')
display(coffee)
```

```python
print(f"Top 5 rows: \n {coffee.head(5)}")
print("|+" * 20)
print(f"Last 5 rows: \n {coffee.tail(5)}")
print("|+" * 20)
print(f"Random 5 rows: \n {coffee.sample(5)}")
```

```python
# using loc[rows, columns] method
# access dataframe using their names

# the stop index is inclusive in loc unlike in iloc
print(f"Units Sold between days 2 to 5: \n{coffee.loc[2:5, 'Units Sold']}")
print("|*" * 20)
print(f"Type of Units Sold between days 2 to 5: \n{coffee.loc[2:5, ['Coffee Type', 'Units Sold']]}")
print("|*" * 20)
```

```python
# using iloc[rows, columns] method
# access dataframe using their names

# the stop index is exclusive in iloc unlike in loc
print(f"Units Sold between days 2 to 5: \n{coffee.iloc[2:6, 0:3]}")
print("|*" * 20)
print(f"Type of Units Sold from days 2: \n{coffee.iloc[2:, 1:3]}")
print("|*" * 20)
```

```python
# modify using loc and iloc
coffee.loc[2, 'Units Sold'] = 25
print(coffee.head(3))
```

```python
coffee.iloc[1:3, 1:3] = ['Mocha', 20]
print(coffee.head(4))
```

* we can also use .at() and .iat() to access values

```python
# Grab colums
print(f"Coffee Types:\n {coffee["Coffee Type"]}")
print(f"Units Sold:\n {coffee["Units Sold"]}")
```

```python
# sort coffee in ascending order
coffee.sort_values('Units Sold', ascending=True)
```

```python
# sort coffee in descending/ascending order based on multiple values
coffee.sort_values(['Units Sold', 'Day'], ascending=[False, True])
```

## Adding/Removing Columns

```python
# add price column to coffee
coffee['price'] = 4.99
print(coffee.head())
```

```python
# add new price column where espresso is 3.99, else 5.99
import numpy as np
coffee['new_price'] = np.where(coffee['Coffee Type'] == "Espresso", 3.99, 5.99)
print(coffee.head())
```

```python
# use drop method to remove column
coffee.drop(columns=["price"], inplace=True)
print(coffee)
```

```python
# calculate revenue add it as new column
coffee['revenue'] = coffee['Units Sold'] * coffee['new_price']
print(coffee.head(5))
```

```python
# rename new_price to Price and revenue to Revenue
coffee.rename(columns={'new_price': 'Price', 'revenue': 'Revenue'}, inplace=True)
print(coffee.head(5))
```

```python
# add column of datetime object
bios_data_new = bios_data.copy()
bios_data_new['datetime'] = pd.to_datetime(bios_data_new['born_date'])
print(bios_data_new.head(3))
```

```python
# create csv for bios_data_new
bios_data_new.to_csv('bios_data_new.csv', index=False)
```

```python
# add heigh category column
bios_data_new['height_cateogry'] = bios_data_new['height_cm'].apply(lambda x: "Short" if x < 165 else "Average" if x < 185 else "Tall")
print(bios_data_new.head(10))
```

```python
# categorize athelete
def categorize_athelete(ath):
    if ath['height_cm'] < 175 and ath['weight_kg'] < 70:
        return "Lightweight"
    elif ath['height_cm'] < 185 and ath['weight_kg'] < 85:
        return "Middleweight"
    else:
        return "Heavyweight"

bios_data_new['category'] = bios_data_new.apply(categorize_athelete, axis=1)
print(bios_data_new.head(5))
```

## Merging andConcatenating Data

```python
# list people playing for country which is not their Place of birth
bios_new[bios_new["NOC_x"] != bios_new["born_country_full"]][['name', 'NOC_x', 'born_country_full']].head(5)
```

```python
# retrieve pakistani olympians
pakistanis = bios_data[bios_data['NOC'] == "Pakistan"]
display(pakistanis.tail(5))
```

## Advanced Functionality

```python
# shift method to get yesterday revenue
coffee["Yesterday Revenue"] = coffee["Revenue"].shift(2)
display(coffee.head(5))
```

```python
coffee["Pct Change"] = coffee["Revenue"] / coffee["Yesterday Revenue"] * 100
display(coffee.head(5))
```

```python
# use rank function
bios_data["height_rank"] = bios_data["height_cm"].rank(ascending=False)
display(bios_data.sort_values("height_rank").head(5))
```

```python
# use cumsum method to get cumulative revenue
coffee["Cumulative Revenue"] = coffee["Revenue"].cumsum()
display(coffee)
```

```python
# using rolling function to have 3 day units sold for latte
latte = coffee[coffee["Coffee Type"] == "Latte"].copy()
latte["3day US"] = latte["Units Sold"].rolling(3).sum()
display(latte)
```
